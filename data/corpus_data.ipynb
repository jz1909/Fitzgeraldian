```python
import re
import numpy as np

text = open("fitzgerald_data/input.txt", encoding="utf-8").read()

paragraphs = [p for p in text.split("\n\n") if p.strip()]
lengths = np.array([len(p) for p in paragraphs])
print(f"{len(paragraphs):,} paragraphs")
```

```python
print(f"longest : {lengths.max():,} chars")
print(f"mean    : {lengths.mean():,.0f} chars")
print(f"median  : {np.median(lengths):,.0f} chars")
for q in [90, 95, 99]:
    print(f"p{q}     : {np.percentile(lengths, q):,.0f} chars")

block_size = 256
print(f"\n{(lengths > block_size).mean():.1%} of paragraphs are longer than block_size={block_size}")
```

```python
# the 10 longest paragraphs
order = np.argsort(lengths)[::-1]
for i in order[:10]:
    print(f"{lengths[i]:>6,}  {paragraphs[i][:100]!r}...")
```

```python
# distribution by length bucket
edges = [e for e in [0, 64, 128, 256, 512, 1024, 2048, 4096] if e < lengths.max()] + [lengths.max() + 1]
counts, _ = np.histogram(lengths, bins=edges)
for lo, hi, n in zip(edges[:-1], edges[1:], counts):
    print(f"{lo:>5}-{hi-1:<6} {n:>6,}  {'#' * int(60 * n / counts.max())}")
```
